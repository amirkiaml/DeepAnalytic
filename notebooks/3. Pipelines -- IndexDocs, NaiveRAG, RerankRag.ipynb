# Splitting the Text

```python
import tiktoken
import re

# Create the length function
def tiktoken_len(text):
    # Get the Encoding object that tiktoken uses for the gpt-4o model
    # (this returns something like: <Encoding 'o200k_base'>)
    input_string = str(tiktoken.encoding_for_model('gpt-4o'))

    # Extract just the encoding name (e.g. 'o200k_base') from that string
    # by grabbing whatever's between the first pair of single quotes
    match = re.search(r"'(.*?)'", input_string)
    result = match.group(1)

    # Load the tokenizer for that encoding name directly
    tokenizer = tiktoken.get_encoding(result)

    # Encode the text into tokens
    # disallowed_special=() means don't treat any special tokens as
    # "disallowed" (avoids errors if the text contains strings that
    # look like special tokens, e.g. "<|endoftext|>")
    tokens = tokenizer.encode(
        text,
        disallowed_special=()
    )

    # Return the number of tokens
    return len(tokens)
```

```python
# The non-interactive splitter set-up

from langchain_text_splitters import RecursiveCharacterTextSplitter

class TextSplitter:
    def __init__(self, chunk_size=400, chunk_overlap=20, separators=None):
        # Use provided separators, or fall back to sensible defaults.
        # (Using None as the default arg instead of a list directly avoids
        # the classic Python "mutable default argument" bug.)
        separators = separators or ["\n\n", "\n", " ", ""]

        # Build the actual LangChain splitter.
        # chunk_size / chunk_overlap are measured in tokens, not characters,
        # because length_function=tiktoken_len below counts tokens.
        self.text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,        # max tokens per chunk
            chunk_overlap=chunk_overlap,  # tokens shared between consecutive chunks
            length_function=tiktoken_len, # defined earlier in the notebook
            separators=separators         # split priority: paragraphs > lines > words > chars
        )

    def split_text(self, text):
        # Splits one document's text into a list of chunk strings.
        return self.text_splitter.split_text(text)
```

```python
# Non iteractively creating embVector DB  


from pinecone import Pinecone, ServerlessSpec
import time
import getpass
import os


class VectorDB:
    """
    Non-interactive Pinecone wrapper — no input() prompts.
    All config is passed as constructor args (with sensible defaults)
    instead of typed in at runtime.
    """

    def __init__(self, cloud="aws", region="us-east-1"):
        # Read the API key from the environment (.env via load_dotenv()).
        # Only fall back to a secure manual prompt if it's genuinely missing.
        api_key = os.getenv("PINECONE_API_KEY")
        if not api_key:
            api_key = getpass.getpass("Enter your Pinecone API key: ")

        # Initialize the Pinecone client — this is your connection to
        # your whole Pinecone account (can list/create/delete any index on it).
        self.pc = Pinecone(api_key=api_key)

        # ServerlessSpec defines *where* a new index gets hosted.
        # aws / us-west-2 is a solid, cheap, widely-available default —
        # only override if you have a specific latency/need.
        self.spec = ServerlessSpec(cloud=cloud, region=region)

        # Populated later by create_index() / connect_to_index()
        self.index_name = None
        self.index = None

    def list_indexes(self):
        # Returns metadata for every index in your Pinecone account.
        return self.pc.list_indexes()

    def list_cloud(self, index_name=None):
        # Prints which cloud/region a given index is configured on.
        index_name = index_name if index_name else self.index_name
        if index_name and index_name in [index['name'] for index in self.list_indexes()]:
            print(f"Index '{index_name}' is configured on:")
            print(f"Cloud Provider: {self.spec.cloud}")
            print(f"Cloud Region: {self.spec.region}")
        else:
            print(f"Index '{index_name}' does not exist.")

    def create_index(self, index_name, dimension=1536, metric="cosine"):
        """
        dimension defaults to 1536 — must match your embedding model's
        output size (text-embedding-3-small = 1536-dim). Mismatched
        dimensions between index and embeddings will error at query time.

        metric defaults to "cosine" — standard for OpenAI embeddings.
        """
        self.index_name = index_name
        existing_indexes = [index['name'] for index in self.list_indexes()]

        if self.index_name not in existing_indexes:
            print(f"Creating index '{self.index_name}'...")
            self.pc.create_index(
                self.index_name,
                dimension=dimension,
                metric=metric,
                spec=self.spec
            )
            # Index creation is async — poll until Pinecone reports it ready.
            while not self.pc.describe_index(self.index_name).status['ready']:
                time.sleep(1)  # bug fix: was self.time.sleep (not a class attr)
            print(f"Index '{self.index_name}' created and is now ready.")
        else:
            print(f"Index '{self.index_name}' already exists. No action taken.")

    def connect_to_index(self, index_name=None):
        # Connects to an existing index and returns the usable Index object.
        index_name = index_name or self.index_name
        if index_name and index_name in [index['name'] for index in self.list_indexes()]:
            self.index_name = index_name
            self.index = self.pc.Index(self.index_name)
            print(f"Connected to index '{self.index_name}'.")
            return self.index
        else:
            raise Exception(f"Index '{index_name}' does not exist.")

    def delete_index(self, index_name=None):
        # Permanently deletes an index and its vectors. No confirmation
        # prompt — call this deliberately, since it's non-interactive.
        index_name = index_name if index_name else self.index_name
        if index_name and index_name in [index['name'] for index in self.list_indexes()]:
            self.pc.delete_index(index_name)
            print(f"Index '{index_name}' has been deleted.")
        else:
            print(f"Index '{index_name}' does not exist.")
```

# The Indexing Pipeline

```python
from tqdm import tqdm
from uuid import uuid4
import pandas as pd  # Assuming you're using a DataFrame

# Define a class to encapsulate the pipeline
class TextProcessingPipeline:
    def __init__(self, dataframe):
        self.df = dataframe
        self.setup_components()

    def setup_components(self):
        print('Setting up the text splitter..')
        print('------------------')
        self.text_splitter = TextSplitter()
        
        print('\nSetting up the embedding model..')
        print('------------------')
        self.embed = EmbedCreator()

        print('\nSetting up the vector database..')
        print('------------------')
        self.index_name = 'sep-articles'
        self.vector_db = VectorDB()
        self.vector_db.create_index(self.index_name)
        self.index = self.vector_db.connect_to_index()

    def process_texts(self, batch_limit=100):
        texts = []
        metadatas = []
        
        # Process each row in the dataframe
        for i in tqdm(range(len(self.df))):
            metadata = {
                'article_id': str(self.df['ID'].iloc[i]),
                'source': self.df['Url'].iloc[i],
                'title': self.df['Title'].iloc[i],
                'authors': list(self.df['Authors'].iloc[i]),  # convert ndarray -> list
                'citation': self.df['BibURL'].iloc[i],
                'date': str(self.df['Date'].iloc[i])
            }
            
            # Split text into chunks and create metadata for each chunk
            record_texts = self.text_splitter.split_text(self.df['Text'].iloc[i])
            record_metadatas = [{'chunk': j, 'text': text, **metadata} for j, text in enumerate(record_texts)]
            
            texts.extend(record_texts)
            metadatas.extend(record_metadatas)
            
            # Check if batch limit is reached to process and upsert data
            if len(texts) >= batch_limit:
                self.embed_and_upsert(texts, metadatas)
                texts = []
                metadatas = []
                
        # Process any remaining texts and metadata after the loop
        if texts:
            self.embed_and_upsert(texts, metadatas)

    def embed_and_upsert(self, texts, metadatas):
        ids = [str(uuid4()) for _ in range(len(texts))]
        embeds = self.embed.embed_documents(texts)
        self.index.upsert(vectors=zip(ids, embeds, metadatas))

# Example of how to use the pipeline
if __name__ == '__main__':
    #df = pd.read_csv('path_to_your_data.csv')  # Load your data into a DataFrame
    pipeline = TextProcessingPipeline(df)
```

```python
#non-interactively retrieving stuff

import os
from dotenv import load_dotenv
from pinecone import Pinecone
from langchain_pinecone import PineconeVectorStore
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_classic.chains import create_retrieval_chain
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate

# Load variables from .env into the environment
load_dotenv()

# This MUST match the embedding model used when the Pinecone index was
# originally populated (in EmbedCreator). Changing this without
# re-embedding the entire index will silently break retrieval quality,
# or throw a dimension-mismatch error from Pinecone.
EMBEDDING_MODEL = 'text-embedding-3-small'


class Naive_RAG:
    def __init__(
        self,
        index_name='sep-articles',   # your existing Pinecone index
        llm_model='gpt-4o',          # solid general-purpose model
        temperature=0.0,             # deterministic answers for QA
        k=5                          # chunks retrieved per query
    ):
        self.index_name = index_name
        self.llm_model = llm_model
        self.temperature = temperature
        self.k = k

        self.pinecone_api_key = self.setup_pinecone_api_key()
        self.openai_api_key = self.setup_openai_api_key()
        self.pc = self.setup_pinecone()
        self.index = self.connect_to_index()
        self.embed = self.load_embedding_model()
        self.vectorstore = PineconeVectorStore(self.index, self.embed)
        self.llm = self.setup_llm()
        self.qa = self.setup_qa()

    def setup_pinecone_api_key(self):
        key = os.environ.get('PINECONE_API_KEY')
        if not key:
            raise ValueError("PINECONE_API_KEY not found in .env file.")
        return key

    def setup_openai_api_key(self):
        key = os.environ.get('OPENAI_API_KEY')
        if not key:
            raise ValueError("OPENAI_API_KEY not found in .env file.")
        return key

    def setup_pinecone(self):
        return Pinecone(api_key=self.pinecone_api_key)

    def connect_to_index(self):
        return self.pc.Index(self.index_name)

    def load_embedding_model(self):
        # Fixed, not configurable — see EMBEDDING_MODEL comment above
        return OpenAIEmbeddings(
            model=EMBEDDING_MODEL,
            openai_api_key=self.openai_api_key
        )

    def setup_llm(self):
        return ChatOpenAI(
            openai_api_key=self.openai_api_key,
            model=self.llm_model,
            temperature=self.temperature
        )

    def setup_qa(self):
        system_prompt = (
            "Use the given context to answer the question. "
            "If you don't know the answer, say you don't know. "
            "Keep the answer concise.\n\n"
            "Context: {context}"
        )
        prompt = ChatPromptTemplate.from_messages([
            ("system", system_prompt),
            ("human", "{input}"),
        ])

        question_answer_chain = create_stuff_documents_chain(self.llm, prompt)
        retriever = self.vectorstore.as_retriever(search_kwargs={'k': self.k})
        return create_retrieval_chain(retriever, question_answer_chain)

    def invoke_qa(self, question):
        result = self.qa.invoke({"input": question})
        return result
    
    def query(self, query_text, k=20):
        """
        Retrieves raw Document objects from the vectorstore without
        generating an LLM answer. Useful as input to a reranker before
        the final generation step.
        """
        return self.vectorstore.similarity_search(query_text, k=k)


# Example usage
if __name__ == '__main__':
    krs = Naive_RAG()
    answer = krs.invoke_qa("What does Socrates think about death?")
    print("\nAnswer:", answer.get("answer"))
```

```python
import os
from dotenv import load_dotenv
import cohere

load_dotenv()

# Init Cohere client — reads key from .env, same pattern as Pinecone/OpenAI
co = cohere.Client(api_key=os.environ.get('COHERE_API_KEY'))


def docs_to_dicts(docs):
    """
    Converts a list of LangChain Document objects (from retriever results)
    into a list of plain dicts, so they can be passed to Cohere's rerank API.
    """
    list_of_dicts = []
    for doc in docs:
        list_of_dicts.append({
            'text': doc.page_content,
            'article_id': doc.metadata['article_id'],
            'authors': doc.metadata['authors'],
            'chunk': doc.metadata['chunk'],
            'citation': doc.metadata['citation'],
            'date': doc.metadata['date'],
            'source': doc.metadata['source'],
            'title': doc.metadata['title']
        })
    return list_of_dicts


def rerank_docs(query, docs, top_n=5, model='rerank-english-v3.0'):
    """
    Takes the raw retriever results (LangChain Document objects) and a query,
    reranks them using Cohere's cross-encoder reranker, and returns the
    top_n most relevant docs (as dicts) sorted by relevance.
    """
    doc_dicts = docs_to_dicts(docs)

    # Cohere's rerank endpoint wants raw text strings to compare against
    # the query, so we pass just the 'text' field per doc.
    texts = [d['text'] for d in doc_dicts]

    response = co.rerank(
        query=query,
        documents=texts,
        top_n=top_n,
        model=model
    )

    # response.results gives back indices into the original `texts` list,
    # sorted by relevance_score descending. Use that to pull the full
    # original dict (with all metadata) back out, plus attach the score.
    reranked = []
    for result in response.results:
        doc = doc_dicts[result.index]
        doc
```
